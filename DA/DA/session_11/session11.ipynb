```python
DROP_COLS = ["PassengerId", 
            "Name", 
            "Ticket"
            ]
import pandas as pd
def drop_cols(df: pd.DataFrame, cols: list[str])->pd.DataFrame: 
    return df.drop(columns= cols)
```

```python
drop_cols(df,DROP_COLS)
```

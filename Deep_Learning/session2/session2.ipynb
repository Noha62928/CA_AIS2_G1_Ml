```python
import pandas as pd 
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
```

```python
data = pd.read_csv('PlayTennis.csv')
data
```

```python
data = pd.read_csv('PlayTennis.csv')
from sklearn.preprocessing import LabelEncoder 
le = LabelEncoder()
for col in data.columns: 
    data[col] = le.fit_transform(data[col])
x = data.drop('Play Tennis', axis = 1)
y = data['Play Tennis']
```

```python
data
```

```python
clf_entropy = DecisionTreeClassifier(criterion='entropy')
clf_entropy.fit(x,y)
y_predict_entropy = clf_entropy.predict(x)
```

```python
from sklearn.metrics import accuracy_score, classification_report
print('acc ', accuracy_score(y, y_predict_entropy))
print('', classification_report,(y, y_predict_entropy))
```

```python
from sklearn import tree 
plt.figure(figsize=(12,8))
tree.plot_tree(clf_entropy, filled=True, feature_names=x.columns, class_names=True)
```

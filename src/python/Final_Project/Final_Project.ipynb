```python
def bankaccount(balance):
    print(f'1. Deposite')
    print(f'2. Withdraw')
    print(f'3. Check balance')
    choice = input(f'Please choose a process: ')
    def Deposite():
        nonlocal balance 
        amount = int(input(f'Please enter amount of deposite: '))
        balance = balance + amount
        print(f'Deposited {amount} and your balance is {balance}')
        return
    def Withdraw():
        nonlocal balance
        amount = int(input(f'Please enter amount of withdraw: '))
        if (amount > balance) :
            print(f'Please enter a valid amount!')
        else : 
            balance = balance - amount
            print(f'Withdraw {amount} and your balance is {balance}')
        return
    def check_balance():
        print(f'Your current balance is {balance}')
        return
    if choice == "1":
        Deposite()
    elif choice == "2":
        Withdraw()
    elif choice == "3":
        check_balance()
    return 
bankaccount(1000)
```

```python
"""
Task 2
calculator  
"""
class Calculator:
    def __init__(self):
        pass

    def Addition(self, x, y):
        return x + y

    def Subtraction(self, x, y):
        return x - y

    def Multiplication(self, x, y):
        return x * y

    def Division(self, x, y):
        return x / y


# Create calculator object
calculator = Calculator()

print("Welcome to the simple calculator")
print("Please select an operation")
print("1. Addition (+)")
print("2. Subtraction (-)")
print("3. Multiplication (*)")
print("4. Division (/)")
print("5. Exit")

user_choice = input("Please enter your choice (1/2/3/4/5): ")

if user_choice == "5":
    print("Thank you, exiting!")

elif user_choice in ["1", "2", "3", "4"]:

    x = float(input("Please enter the first number: "))
    y = float(input("Please enter the second number: "))

    if user_choice == "1":
        print(f"{x} + {y} = {calculator.Addition(x, y)}")

    elif user_choice == "2":
        print(f"{x} - {y} = {calculator.Subtraction(x, y)}")

    elif user_choice == "3":
        print(f"{x} * {y} = {calculator.Multiplication(x, y)}")

    elif user_choice == "4":

        if y == 0:
            print("Cannot divide by zero!")

        else:
            print(f"{x} / {y} = {calculator.Division(x, y)}")

else:
    print("Invalid choice!")
```

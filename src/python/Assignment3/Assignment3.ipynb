```python
def multiplication_table(num):
    for i in range(1, 13):
        multiplication = num * i 
        print(f"{num} * {i} = {multiplication}")
    return 
```

```python
num = int(input("Please enter a number: "))
multiplication_table(num)
```

```python
def twin_primes():
    print(f"Twin Prime:")
    for n in range(0, 999):
        if n % 2 != 0:
            prime1 = True
            for i in range(2, n):
                if n % i == 0:
                    prime1 = False
                    break
            prime2 = True
            for i in range(2, n + 2):
                if (n + 2) % i == 0:
                    prime2 = False
                    break
            if prime1 == True and prime2 == True:
                print(n, "and", n + 2)
    return
```

```python
twin_primes()
```

```python
def prime_factors(num):
    for i in range (2, num):
        while num%i == 0 : 
            n = i 
            num = num/i 
            print(n)
    return
```

```python
prime_factors(56)
```

```python
def decToBin(num):
 n_binary = ""
 for num in range(num, 0, -1): 
  while num > 0:
   n = num%2 
   n_binary = str(n) + n_binary 
   num = int(num/2) 
  print(n_binary)
  return(n_binary)
```

```python
def perfectNums(m, n): 
 for i in range (m, n+1, 1): 
   add = 0         
   for k in range (1, i):             
      if i%k == 0:                 
       add = add + k         
   if add == i:           
    print(i)     
 return
```

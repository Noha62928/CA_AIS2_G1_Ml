```python
def multiplication_table(num):
    for i in range(1, 13):
        multiplication = num * i 
        print(f"{num} * {i} = {multiplication}")
    return 
```

```python
num = int(input("Please enter a number: "))
multiplication_table(num)
```

```python
def twin_primes():
    print(f"Twin Prime:")
    for n in range(0, 999):
        if n % 2 != 0:
            prime1 = True
            for i in range(2, n):
                if n % i == 0:
                    prime1 = False
                    break
            prime2 = True
            for i in range(2, n + 2):
                if (n + 2) % i == 0:
                    prime2 = False
                    break
            if prime1 == True and prime2 == True:
                print(n, "and", n + 2)
    return
```

```python
twin_primes()
```

```python
class multiplication:
   """
   Task 1 : inputs a number and prints the multiplication table 
 
   """
   def __init__(self, num) -> None:
    self.num = num
   def multiplication_table(self):
     for i in range(1, 13):
             multiplication = self.num * i 
             print(f"{self.num} * {i} = {multiplication}")
     return
```

```python
"""
Enter a number
"""
num = int(input("Please enter a number!"))
"""
create an object 
"""
m1 = multiplication(num)
m1.multiplication_table()
```

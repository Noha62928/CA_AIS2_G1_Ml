```python
class Shape:
    def calculate_area(self):
        pass

class Circle(Shape):
    def __init__(self, radius) -> None:
        self.radius = radius

    def calculate_area(self): 
        return 3.14 * self.radius * self.radius 
    
class Square(Shape):
    def __init__(self, side) -> None:
        self.side = side 

    def calculate_area(self): 
        return self.side ** 2

if __name__ == "__main__":
    x = Circle(5)
    y = Square(4)
    print("Area Circle: ", x.calculate_area())
    print("Area Square: ", y.calculate_area())
```
